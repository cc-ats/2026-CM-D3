```python
# ============================================================
# SCATTER PLOT: D3(BJ) vs D4 ENERGIES
# MATCHED TO REWEIGHTING-ENTROPY FIGURE FORMATTING
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from matplotlib.ticker import FormatStrFormatter,MultipleLocator,AutoMinorLocator
```

```python
# ============================================================
# GENERAL PLOT SETTINGS
# ============================================================

# Keep text editable when the figure is opened in Illustrator
plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42

plt.rcParams["axes.linewidth"] = 0.8
plt.rcParams["axes.grid"] = False
plt.rcParams["axes.xmargin"] = 0.05
plt.rcParams["axes.ymargin"] = 0.10
```

```python
# ============================================================
# LOAD ENERGIES
# ============================================================

ene_b3lyp_D3 = np.load("d3bj_energy.npy") * 627.5
ene_b3lyp_D4 = np.load("d4_energy.npy") * 627.5

x = np.asarray(ene_b3lyp_D3).flatten()
y = np.asarray(ene_b3lyp_D4).flatten()

print("D3(BJ) shape:", x.shape)
print("D4 shape:", y.shape)

if x.shape != y.shape:
    raise ValueError("D3(BJ) and D4 energy arrays do not have the same shape.")
```

```python
# ============================================================
# FIGURE SIZE
# ============================================================


FIGURE_WIDTH = 4.75
FIGURE_HEIGHT = 4.25


# ============================================================
# EXACT PLOTTING-AREA DIMENSIONS
# ============================================================

PLOT_LEFT = 0.72
PLOT_BOTTOM = 0.62
PLOT_WIDTH = 3.55
PLOT_HEIGHT = 3.15


# ============================================================
# CREATE EXACT-SIZE FIGURE
# ============================================================

fig = plt.figure(
    figsize=(FIGURE_WIDTH, FIGURE_HEIGHT),
    facecolor="white"
)

ax = fig.add_axes([
    PLOT_LEFT / FIGURE_WIDTH,
    PLOT_BOTTOM / FIGURE_HEIGHT,
    PLOT_WIDTH / FIGURE_WIDTH,
    PLOT_HEIGHT / FIGURE_HEIGHT
])


# ============================================================
# COLORS
# ============================================================

cmap = plt.get_cmap("tab20")
POINT_COLOR = cmap(0)   # blue
# ============================================================
# SCATTER PLOT
# ============================================================

ax.scatter(
    x,
    y,
    s=10,
    color=POINT_COLOR,
    alpha=0.45,
    edgecolors="none",
    rasterized=True
)


# ============================================================
# y = x REFERENCE LINE
# ============================================================

xy_min = min(np.min(x), np.min(y))
xy_max = max(np.max(x), np.max(y))
padding = 0.04 * (xy_max - xy_min)

axis_min = xy_min - padding
axis_max = xy_max + padding

ax.plot(
    [axis_min, axis_max],
    [axis_min, axis_max],
    linestyle="--",
    linewidth=1.0,
    color="black"
)


# ============================================================
# AXIS LABELS
# ============================================================

ax.set_xlabel(r"D3(BJ) Energy (kcal/mol)")
ax.set_ylabel(r"D4 Energy (kcal/mol)")


# ============================================================
# AXIS LIMITS
# ============================================================

ax.set_xlim(axis_min, axis_max)
ax.set_ylim(axis_min, axis_max)
ax.set_aspect("equal", adjustable="box")
ax.xaxis.set_major_formatter(FormatStrFormatter('%.1f'))
ax.yaxis.set_major_formatter(FormatStrFormatter('%.1f'))
ax.xaxis.set_major_locator(MultipleLocator(0.5))
ax.yaxis.set_major_locator(MultipleLocator(0.5))


# ============================================================
# TICKS
# ============================================================

ax.tick_params(
    axis="both",
    which="major",
    direction="out",
    length=4,
    width=0.8
)

ax.tick_params(
    axis="both",
    which="minor",
    direction="out",
    length=2,
    width=0.6
)

ax.xaxis.set_minor_locator(AutoMinorLocator(2))
ax.yaxis.set_minor_locator(AutoMinorLocator(2))



# ============================================================
# SAVE FIGURE
# ============================================================

fig.savefig(
    "D3BJ_vs_D4_scatter_matched_format.pdf",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
